### Example 1: Simple Streamlit App

Let's create our first app!

````python
# calculator_app.py
import streamlit as st

st.set_page_config(page_title="Calculator", page_icon="🧮", layout="centered")
st.title("🧮 Calculator")

# Two-column layout for inputs
col1, col2 = st.columns(2)

with col1:
    num1 = st.number_input("First number", value=0.0, step=1.0)

with col2:
    num2 = st.number_input("Second number", value=0.0, step=1.0)

# Operation selector
operation = st.selectbox(
    "Operation",
    ["+", "-", "*", "/"],
    help="Choose the arithmetic operation",
)

# Calculate button
if st.button("Calculate", type="primary", use_container_width=True):
    try:
        if operation == "+":
            result = num1 + num2
        elif operation == "-":
            result = num1 - num2
        elif operation == "*":
            result = num1 * num2
        elif operation == "/":
            if num2 == 0:
                st.error("❌ Cannot divide by zero")
                st.stop()
            result = num1 / num2

        st.success(f"**Result:** {num1} {operation} {num2} = **{result}**")
        st.metric("Result", result)

    except Exception as e:
        st.error(f"Error: {e}")
````

## Part 2: 💾 Session State (Memory)

### The Problem

Streamlit reruns your entire script on every interaction. Variables get reset!

```python
count = 0  # This resets to 0 every time!
if st.button('Count'):
    count += 1
    st.write(count)  # Always shows 1
```

### The Solution: Session State

`st.session_state` persists data across reruns!

````python
# Session state demo code

SESSION_STATE_APP = '''
import streamlit as st

# Initialize session state
if 'count' not in st.session_state:
    st.session_state.count = 0

if 'messages' not in st.session_state:
    st.session_state.messages = []

st.title('💾 Session State Demo')

# Counter that persists
if st.button('Increment'):
    st.session_state.count += 1

st.write(f'Count: {st.session_state.count}')

# Chat history that persists
user_msg = st.text_input('Your message:')
if st.button('Send') and user_msg:
    st.session_state.messages.append({'role': 'user', 'content': user_msg})
    st.session_state.messages.append({'role': 'bot', 'content': f'Echo: {user_msg}'})

# Display chat history
for msg in st.session_state.messages:
    with st.chat_message(msg['role']):
        st.write(msg['content'])
'''

print('💾 Session State Pattern:')
print(SESSION_STATE_APP)
````

````python
# Complete Digital Twin Streamlit App

DIGITAL_TWIN_APP = '''
import streamlit as st
from datetime import datetime
import time

# Page config (must be first Streamlit command)
st.set_page_config(
    page_title="Digital Twin Chat",
    page_icon="🤖",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Initialize session state
if 'messages' not in st.session_state:
    st.session_state.messages = []
    # Add welcome message
    st.session_state.messages.append({
        'role': 'assistant',
        'content': 'Hello! I\'m your Digital Twin. How can I help you today?',
        'timestamp': datetime.now()
    })

if 'user_name' not in st.session_state:
    st.session_state.user_name = ''

# Simple response function (replace with your LLM later)
def generate_response(user_message: str) -> str:
    '''Generate a response from the digital twin.'''
    # Placeholder - integrate your Week 1-3 digital twin here!
    responses = {
        'hello': 'Hi there! How are you doing?',
        'help': 'I can chat with you, remember our conversation, and use tools!',
        'time': f'The current time is {datetime.now().strftime("%H:%M:%S")}',
    }

    for key, response in responses.items():
        if key in user_message.lower():
            return response

    return f"You said: {user_message}. I'm learning to respond better!"

# Sidebar
with st.sidebar:
    st.title('⚙️ Settings')

    # User profile
    st.subheader('Profile')
    name = st.text_input('Your Name:', st.session_state.user_name)
    if name != st.session_state.user_name:
        st.session_state.user_name = name

    st.divider()

    # Stats
    st.subheader('📊 Stats')
    st.metric('Messages', len(st.session_state.messages))
    st.metric('User Messages',
              len([m for m in st.session_state.messages if m['role'] == 'user']))

    st.divider()

    # Clear chat
    if st.button('🗑️ Clear Chat', use_container_width=True):
        st.session_state.messages = []
        st.rerun()

# Main chat interface
st.title('🤖 Digital Twin Chat')

if st.session_state.user_name:
    st.write(f'Chatting as: **{st.session_state.user_name}**')

# Chat container
chat_container = st.container()

# Display all messages
with chat_container:
    for message in st.session_state.messages:
        with st.chat_message(message['role']):
            st.write(message['content'])
            if 'timestamp' in message:
                st.caption(message['timestamp'].strftime('%H:%M:%S'))

# Chat input
user_input = st.chat_input('Type your message...')

if user_input:
    # Add user message
    st.session_state.messages.append({
        'role': 'user',
        'content': user_input,
        'timestamp': datetime.now()
    })

    # Display user message immediately
    with chat_container:
        with st.chat_message('user'):
            st.write(user_input)

    # Generate and display response
    with chat_container:
        with st.chat_message('assistant'):
            with st.spinner('Thinking...'):
                time.sleep(0.5)  # Simulate processing
                response = generate_response(user_input)
                st.write(response)

    # Add assistant message to history
    st.session_state.messages.append({
        'role': 'assistant',
        'content': response,
        'timestamp': datetime.now()
    })

    # Force rerun to show new messages
    st.rerun()
'''

print('🎭 Complete Digital Twin Chat UI:')
print('\n💡 Save as digital_twin_app.py and run:')
print('   streamlit run digital_twin_app.py')
print('\n📝 Code preview:')
print(DIGITAL_TWIN_APP[:500] + '...\n[Full code above]')
````

````python
# todo_app.py
import streamlit as st

st.set_page_config(page_title="Todo List", page_icon="✅", layout="centered")
st.title("✅ Todo List")

# Initialize session state
if "todos" not in st.session_state:
    st.session_state.todos = []

# --- Add todo ---
with st.form("add_todo", clear_on_submit=True):
    new_todo = st.text_input("New task", placeholder="What needs doing?")
    submitted = st.form_submit_button("Add", use_container_width=True)

    if submitted:
        if new_todo.strip():
            st.session_state.todos.append({
                "text": new_todo.strip(),
                "done": False,
            })
        else:
            st.warning("Please enter a task.")

# --- Display todos ---
st.divider()
st.subheader(f"Tasks ({len(st.session_state.todos)})")

if not st.session_state.todos:
    st.info("No tasks yet. Add one above!")
else:
    for i, todo in enumerate(st.session_state.todos):
        col1, col2, col3 = st.columns([0.1, 0.8, 0.1])

        with col1:
            checked = st.checkbox("", value=todo["done"], key=f"done_{i}")
            if checked != todo["done"]:
                st.session_state.todos[i]["done"] = checked
                st.rerun()

        with col2:
            style = "text-decoration: line-through; opacity: 0.5;" if todo["done"] else ""
            st.markdown(f'<div style="{style}">{todo["text"]}</div>', unsafe_allow_html=True)

        with col3:
            if st.button("🗑️", key=f"del_{i}"):
                st.session_state.todos.pop(i)
                st.rerun()

    # --- Clear all ---
    st.divider()
    col1, col2 = st.columns([0.7, 0.3])
    with col2:
        if st.button("Clear All", type="secondary", use_container_width=True):
            st.session_state.todos = []
            st.rerun()

# --- Stats ---
done_count = sum(1 for t in st.session_state.todos if t["done"])
st.caption(f"✅ {done_count} done · ⏳ {len(st.session_state.todos) - done_count} pending")
````

````python
# digital_twin_app.py
import streamlit as st
from datetime import datetime
import time
import json

# ============================================================
# PAGE CONFIG (must be first Streamlit call)
# ============================================================
st.set_page_config(
    page_title="Digital Twin",
    page_icon="🤖",
    layout="wide",
    initial_sidebar_state="expanded",
)

# ============================================================
# SESSION STATE
# ============================================================
def init_state():
    defaults = {
        "messages": [{
            "role": "assistant",
            "content": "Hello! I'm your Digital Twin. How can I help today?",
            "timestamp": datetime.now(),
        }],
        "user_name": "Guest",
        "temperature": 0.7,
        "use_tools": True,
        "show_debug": False,
        "error_count": 0,
    }
    for key, value in defaults.items():
        if key not in st.session_state:
            st.session_state[key] = value

init_state()

# ============================================================
# BACKEND (mock — swap for your real DigitalTwin)
# ============================================================
def generate_response(user_message: str) -> dict:
    """
    In production, replace with:
        - Week 1: LLM call via GroqClient
        - Week 2: RAG retrieve() + ask()
        - Week 3: tool dispatch + guardrails
    """
    msg_lower = user_message.lower()

    # Simulate tool routing
    tool_used = None
    if any(w in msg_lower for w in ["time", "date", "day"]):
        tool_used = "get_time"
        response = f"It's {datetime.now().strftime('%H:%M')} on {datetime.now().strftime('%A, %B %d, %Y')}."
    elif any(w in msg_lower for w in ["calculate", "+", "-", "*", "/"]) and any(c.isdigit() for c in user_message):
        tool_used = "calculator"
        import re
        m = re.search(r"-?\d[\d\s+\-*/().]*", user_message)
        try:
            result = eval(m.group(0).strip().rstrip("?"), {"__builtins__": {}}, {})
            response = f"The result is **{result}**."
        except Exception:
            response = "I couldn't parse that expression."
    elif "remember" in msg_lower:
        tool_used = "remember"
        response = "Got it. I'll remember that."
    elif any(w in msg_lower for w in ["hello", "hi", "hey"]):
        response = f"Hi {st.session_state.user_name}! What can I do for you?"
    else:
        response = f"You said: *{user_message}*. I'm still learning!"

    return {"content": response, "tool": tool_used, "temperature": st.session_state.temperature}

# ============================================================
# SIDEBAR
# ============================================================
with st.sidebar:
    st.title("🤖 Digital Twin")
    st.caption("Week 4.1 Capstone")

    # --- Profile ---
    st.subheader("👤 Profile")
    name = st.text_input("Your name", value=st.session_state.user_name)
    if name != st.session_state.user_name:
        st.session_state.user_name = name

    st.divider()

    # --- Settings ---
    st.subheader("⚙️ Settings")
    st.session_state.temperature = st.slider(
        "Temperature", 0.0, 1.0, st.session_state.temperature,
        help="Higher = more creative",
    )
    st.session_state.use_tools = st.toggle("Enable tools", value=st.session_state.use_tools)
    st.session_state.show_debug = st.toggle("Show debug info", value=st.session_state.show_debug)

    st.divider()

    # --- Stats ---
    st.subheader("📊 Stats")
    user_msgs = sum(1 for m in st.session_state.messages if m["role"] == "user")
    asst_msgs = sum(1 for m in st.session_state.messages if m["role"] == "assistant")
    c1, c2 = st.columns(2)
    c1.metric("You", user_msgs)
    c2.metric("Twin", asst_msgs)
    st.metric("Errors", st.session_state.error_count)

    st.divider()

    # --- Actions ---
    col_a, col_b = st.columns(2)
    with col_a:
        if st.button("🗑️ Clear", use_container_width=True):
            st.session_state.messages = [{
                "role": "assistant",
                "content": "Chat cleared. Fresh start!",
                "timestamp": datetime.now(),
            }]
            st.rerun()
    with col_b:
        export = json.dumps(
            [{**m, "timestamp": m["timestamp"].isoformat()} for m in st.session_state.messages],
            indent=2,
        )
        st.download_button(
            "⬇️ Export", export,
            file_name=f"chat_{datetime.now():%Y%m%d_%H%M}.json",
            mime="application/json",
            use_container_width=True,
        )

# ============================================================
# MAIN AREA
# ============================================================
st.title("🤖 Digital Twin Chat")
st.caption(f"Chatting as **{st.session_state.user_name}** "
           f"· temperature={st.session_state.temperature:.1f} "
           f"· tools={'on' if st.session_state.use_tools else 'off'}")

# --- Chat history ---
for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.write(msg["content"])
        st.caption(msg["timestamp"].strftime("%H:%M:%S"))
        if st.session_state.show_debug and msg.get("tool"):
            st.code(f"tool_used: {msg['tool']}", language="text")

# --- Input ---
if user_input := st.chat_input("Type your message..."):
    # User bubble
    st.session_state.messages.append({
        "role": "user",
        "content": user_input,
        "timestamp": datetime.now(),
    })
    with st.chat_message("user"):
        st.write(user_input)

    # Assistant bubble
    with st.chat_message("assistant"):
        with st.spinner("Thinking..."):
            try:
                t0 = time.perf_counter()
                result = generate_response(user_input)
                elapsed = time.perf_counter() - t0
            except Exception as e:
                st.session_state.error_count += 1
                result = {"content": f"Sorry, something went wrong: {e}", "tool": None}
                elapsed = 0.0

        st.write(result["content"])
        if st.session_state.show_debug:
            st.caption(f"⏱ {elapsed*1000:.0f}ms · tool={result['tool']}")

    st.session_state.messages.append({
        "role": "assistant",
        "content": result["content"],
        "tool": result["tool"],
        "timestamp": datetime.now(),
    })
````
